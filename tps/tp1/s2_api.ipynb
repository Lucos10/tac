## Je vais utiliser [OCR API](https://ocr.space/ocrapi) pour extraire les textes des PDFs de CAMille.

```python
# Le code a été adapté à partir de https://ocr.space/ocrapi#python

def ocr_space_url(url, overlay=False, api_key='API_KEY'):
    payload = {
        'url': url,
        'isOverlayRequired': overlay,
    }

    headers = {'apikey': api_key}

    r = requests.post(
        'https://api.ocr.space/parse/image',
        data=payload,
        headers=headers
    )

    return r.content.decode()

# La plus part des PDF du site sont trop grands pour l'API

test_url = ocr_space_url(
    url='https://max.de.wilde.web.ulb.be/camille/KB_JB421_1899-05-15_01-00003.pdf',
    api_key='K8525569979629888957'
)

print(test_url)
```
